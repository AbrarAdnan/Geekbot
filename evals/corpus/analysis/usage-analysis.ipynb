# Lantern pilot usage analysis

Prepared by Daniel Reyes on 5 May 2025. The data is the usage export for the four-week pilot, 7 April to 2 May 2025, with four pilot users.

## Headline numbers

- The four pilot users sent 1,386 prompts in total, an average of 69 per working day over the 20 working days.
- Tuesday was the busiest weekday and the hour from 10:00 to 11:00 was the busiest hour.
- The median time to the first word of an answer was 2.1 seconds in real use. That is slower than the 1.8 seconds of the 31 March benchmark because people uploaded documents, which makes the model read more before it answers.
- The satisfaction survey gave 4.3 out of 5.

## What people used it for

Each prompt carries a task label that the user picks from a short list. The shares of all prompts were:

- summarising meeting notes: 41 percent
- drafting client emails: 27 percent
- searching project documents: 19 percent
- everything else: 13 percent

```python
import pandas as pd

usage = pd.read_csv("usage_export.csv", parse_dates=["timestamp"])
usage["hour"] = usage["timestamp"].dt.hour
usage["weekday"] = usage["timestamp"].dt.day_name()

task_share = usage["task"].value_counts(normalize=True).round(2)
busiest_hour = usage["hour"].value_counts().idxmax()
busiest_weekday = usage["weekday"].value_counts().idxmax()
```

## Caveats

- Hannah Frost chose the pilot users, and all four are heavy computer users, so the numbers are probably higher than the studio average.
- The export holds only the task label, never the text of a prompt, because the audit log does not store it.
- Four people is a small sample. The satisfaction score should be read as a mood, not a measurement.

```python
# Median response time per week, to see whether uploads slow things down
weekly_median = usage.groupby(usage["timestamp"].dt.isocalendar().week)["first_token_seconds"].median()
```
